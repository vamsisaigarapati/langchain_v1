### Groq

```python
# Cell 1: Check Groq API Access
import os
from dotenv import load_dotenv
from groq import Groq

load_dotenv()

client = Groq(api_key=os.getenv("GROQ_API_KEY"))

# List available models
models = client.models.list()
print(f"✅ Total Models Available: {len(models.data)}\n")

for model in models.data:
    if hasattr(model, 'context_window') and model.context_window > 100000:
        print(f"📌 {model.id}")
        print(f"   Context: {model.context_window:,} tokens\n")
```

```python
import os
from langchain.chat_models import init_chat_model
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
model = init_chat_model("groq:meta-llama/llama-guard-4-12b")
response = model.invoke("Why do parrots talk?")
response
```

```python
for chunk in model.stream("write a 300 word paragraph on the topic of 'The benefits of using a streamlit app for data analysis'"):
    print(chunk.text)
```

### Batch Inference

```python
responses = model.batch(["Hello, how are you?", "I am a language model.",
"What is the capital of France?", "What is the meaning of life?"
,"What is the square root of 2?"
])
for response in responses:
    print(response.text)
```
